```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
df = pd.read_csv(r"C:\Users\shris\Downloads\covid_19 (1).csv")

df.head()
```

Countplot "How many of each category?"

```python
sns.countplot(
    data=df,
    x="continent",
    order=df["continent"].value_counts().index
)

plt.xticks(rotation=45)
plt.title("COVID-19 Records by Continent")
plt.xlabel("Continent")
plt.ylabel("Count")
plt.show()
```

histplot

```python

sns.histplot(data=df, x="Deaths", bins=10, kde=True)

plt.title("Distribution of COVID-19 Deaths")
plt.xlabel("Number of Deaths")
plt.ylabel("Frequency")
plt.show()
```

box plot

```python
covid_df = df[df["continent"].notna()]
covid_df = covid_df[covid_df["continent"] != "All"]

sns.boxplot(data=covid_df, x="continent", y="Cases")

plt.title("COVID-19 Cases by Continent")
plt.xlabel("Continent")
plt.ylabel("Number of Cases")
plt.xticks(rotation=45)
plt.show()
```

BARPLOT

```python
covid_df = df[df["continent"].notna()]
covid_df = covid_df[covid_df["continent"] != "All"]

sns.barplot(data=covid_df, x="continent", y="Cases")

plt.title("Average COVID-19 Cases by Continent")
plt.xlabel("Continent")
plt.ylabel("Average Cases")
plt.xticks(rotation=45)
plt.show()
```

Scatterplot

```python
sns.scatterplot(data=df, x="population", y="Cases", hue="continent")

plt.title("Population vs COVID-19 Cases by Continent")
plt.xlabel("Population")
plt.ylabel("COVID-19 Cases")
plt.show()
```

Heatmap

```python
corr = df[["population", "Cases", "Recovered", "Deaths", "Tests"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of COVID-19 Data")
plt.show()
```

PAIRPLOT

```python
sns.pairplot(
    covid_df,
    vars=["Cases", "Recovered", "Deaths", "Tests"],
    hue="continent"
)

plt.show()
```
