# Statistics Tests – A/B Testing Data

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.stats.proportion import proportions_ztest
```

```python
data = pd.read_csv("ab_test.csv")
data.head()
```

```python
print(data.shape)
print(data.info())
print(data.isnull().sum())
```

## Mean, Median and Standard Deviation

```python
print("Mean:", data["converted"].mean())
print("Median:", data["converted"].median())
print("Standard deviation:", data["converted"].std())
```

## 1. One-Sample T-Test

```python
sample = data["converted"]

t_stat, p_value = stats.ttest_1samp(sample, 0.5)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

## 2. Independent T-Test

```python
control = data[data["group"] == "control"]["converted"]
treatment = data[data["group"] == "treatment"]["converted"]

t_stat, p_value = stats.ttest_ind(control, treatment)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

## 3. Chi-Square Test

```python
table = pd.crosstab(data["group"], data["converted"])

chi2, p_value, dof, expected = stats.chi2_contingency(table)

print("Chi-square:", chi2)
print("p-value:", p_value)
```

## 4. A/B Test – Two-Proportion Z-Test

```python
control = data[data["group"] == "control"]
treatment = data[data["group"] == "treatment"]

success = [
    control["converted"].sum(),
    treatment["converted"].sum()
]

total = [
    len(control),
    len(treatment)
]

z_stat, p_value = proportions_ztest(success, total)

print("Z-statistic:", z_stat)
print("p-value:", p_value)
```

## 5. Pearson Correlation Test

```python
correlation, p_value = stats.pearsonr(
    data["user_id"],
    data["converted"]
)

print("Correlation:", correlation)
print("p-value:", p_value)
```

## 6. Shapiro-Wilk Normality Test

```python
stat, p_value = stats.shapiro(
    data["converted"].sample(5000, random_state=42)
)

print("Statistic:", stat)
print("p-value:", p_value)
```

## 7. Histogram

```python
plt.hist(data["converted"], bins=2)

plt.xlabel("Converted")
plt.ylabel("Number of Users")
plt.title("Conversion Distribution")

plt.show()
```

## 8. Box Plot – Control vs Treatment

```python
sns.boxplot(
    x="group",
    y="converted",
    data=data
)

plt.title("Control vs Treatment")
plt.show()
```

## 9. Conversion Rate

```python
conversion_rate = data.groupby("group")["converted"].mean()

print(conversion_rate)
```

## 10. Simple Significance Check

```python
if p_value< 0.05:
    print("Significant result")
else:
    print("Not a significant result")
```
